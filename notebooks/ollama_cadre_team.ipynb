# Cadre team-of-teams with Ollama

This notebook launches three independent Ollama workers for each Cadre role, synthesizes each role team, and optionally creates a final integrated handoff. Keep Ollama running locally before executing the notebook.

```python
from pathlib import Path
import sys

repo = Path.cwd()
if not (repo / 'tools' / 'ollama_cadre_runtime.py').exists():
    repo = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'tools' / 'ollama_cadre_runtime.py').exists())
sys.path.insert(0, str(repo / 'tools'))
from ollama_cadre_runtime import CADRE_ROLES, run_teams, call_ollama, final_prompt

WORKER_MODEL = 'llama3.2:1b'
SYNTHESIS_MODEL = 'qwen3:8b'
TEAM_CONCURRENCY = 10  # use 3-5 if the machine becomes CPU/GPU bound
HOST = 'http://127.0.0.1:11434'
GOAL = 'Improve the portfolio website for new-grad ML Engineer, AI Engineer, Software Engineer, and technical Data Science recruiting.'
```

```python
team_outputs, handoffs = run_teams(GOAL, CADRE_ROLES, HOST, WORKER_MODEL, SYNTHESIS_MODEL, TEAM_CONCURRENCY)
for role, team in team_outputs.items():
    print(f'{role}: {len(team["workers"])} workers complete')
```

```python
final_report = call_ollama(HOST, SYNTHESIS_MODEL, final_prompt(GOAL, handoffs))
print(final_report)
```
