# Cadre team-of-teams with OpenAI

This notebook launches three low-cost OpenAI workers per Cadre role in parallel, synthesizes each team, and creates an optional final handoff. Set `OPENAI_API_KEY` in the environment before running.

```python
from pathlib import Path
import os, sys
repo = Path.cwd()
if not (repo / 'tools' / 'openai_cadre_runtime.py').exists():
    repo = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'tools' / 'openai_cadre_runtime.py').exists())
sys.path.insert(0, str(repo / 'tools'))
from openai_cadre_runtime import CADRE_ROLES, run_teams, call_openai, final_prompt

API_KEY = os.environ['OPENAI_API_KEY']
BASE_URL = 'https://api.openai.com/v1'
WORKER_MODEL = 'gpt-5-mini'
HEAD_MODEL = 'gpt-5-mini'
FINAL_MODEL = 'gpt-5-mini'
TEAM_CONCURRENCY = 5
WORKERS_PER_TEAM = 3
DEBATE_ROUNDS = 1
GOAL = 'Improve Connor van Herick portfolio website for new-grad ML Engineer and Software Engineer recruiting.'
```

```python
team_outputs, handoffs = run_teams(GOAL, CADRE_ROLES, API_KEY, BASE_URL, WORKER_MODEL, HEAD_MODEL, TEAM_CONCURRENCY, WORKERS_PER_TEAM, DEBATE_ROUNDS)
for role, team in team_outputs.items():
    print(f'{role}: {len(team["workers"])} workers complete')
```

```python
final_report = call_openai(API_KEY, BASE_URL, FINAL_MODEL, final_prompt(GOAL, handoffs), max_output_tokens=1200)
print(final_report)
```
